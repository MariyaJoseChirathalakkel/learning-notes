```python
# here is a mathematical expression that takes 3 inputs and produces one output
from math import sin, cos

def f(a, b, c):
  return -a**3 + sin(3*b) - 1.0/c + b**2.5 - a**0.5

print(f(2, 3, 4))
```

Here the equation we have is 

f(a,b,c) = -a**3 + sin(3*b) - 1.0/c + b**2.5 - a**0.5


1. derivative of the function with respect to a can be calculated using power rule  

        derivatives with respect to b and c are considerd as constants, hence this will be 0


        df / da = 3 * -a ** 2 - 0.5 * a ** -0.5
                = 3 * -a ** 2 - 0.5 * 1/a**0.5


2. derivative of the function with respect to b (a nd c will be constant, hence 0)

        using chain rule of calculus sin(3b) = cos(3b) and 3b = 3

        df / db = 3 * cos(3b) + 2.5 * b ** 1.5



3. derivative of the function with respect to c (a,b are constants hence 0)

        df / dc = df / dc  - (1* c ** -1)
                = -1 * -1 * c ** -2
                = 1 * c ** -2
                = 1 / c ** 2

# Section 2: Softmax, Negative Log-Likelihood Loss, and Extended Autograd Operations

## 1. Softmax Function
The softmax function converts a vector of raw, unconstrained model outputs (logits) into a probability distribution. The mathematical formulation involves two primary phases:
* **Exponentiation:** Each logit is exponentiated ($e^x$), which transforms all values into strictly positive numbers and amplifies the relative differences between inputs.
* **Normalization:** Each exponentiated value is divided by the sum of all exponentiated values across the entire vector. This guarantees that every output probability resides between $0$ and $1$, and that the sum of all probabilities equals exactly $1.0$.

## 2. Negative Log-Likelihood (NLL) Loss
Negative Log-Likelihood Loss measures the performance of a classification model by evaluating the predicted probability assigned to the true target class. 
* The operation extracts the probability corresponding to the correct category index.
* The natural logarithm of this probability is computed. Because log probabilities of values less than $1$ are negative, a negative sign is applied to yield a positive loss value.
* High confidence in the correct prediction results in a loss approaching zero, whereas incorrect predictions with low assigned probabilities produce a rapidly increasing loss penalty.

## 3. Explanation of Additional Autograd Operations
To support softmax and NLL loss within an automatic differentiation engine, the foundational computational node class must be extended with specialized mathematical operations. Each operation requires both a forward pass calculation and a backward pass implementation governed by the chain rule:

* **Exponents (`exp`):** Necessary for computing the exponential component of softmax. The backward pass propagates gradients using the fundamental property that the derivative of $e^x$ is $e^x$ itself.

* **Powers (`__pow__`) and Division (`__truediv__`):** Required during the normalization phase of softmax, where individual exponentiated values are divided by the aggregate denominator sum. Division is implemented via negative exponentiation, utilizing the power rule derivative ($\frac{d}{dx} x^n = n \cdot x^{n-1}$) to route gradient signals through fractions.

* **Natural Logarithms (`log`):** Essential for evaluating the log-likelihood loss. The backward pass applies the standard derivative of the natural logarithm ($\frac{1}{x}$) to propagate error signals backward through the logarithmic transformation.

* **Negation (`__neg__`):** Utilized by the negative sign preceding the loss formulation. The backward pass applies a multiplier of $-1.0$ to invert gradient signs correctly through the negation node.

```python
# Value class starter code, with many functions taken out
from math import exp, log

class Value:

  def __init__(self, data, _children=(), _op='', label=''):
    self.data = data
    self.grad = 0.0
    self._backward = lambda: None
    self._prev = set(_children)
    self._op = _op
    self.label = label

  def __repr__(self):
    return f"Value(data={self.data})"

  def __add__(self, other): # exactly as in the video
    other = other if isinstance(other, Value) else Value(other)
    out = Value(self.data + other.data, (self, other), '+')

    def _backward():
      self.grad += 1.0 * out.grad
      other.grad += 1.0 * out.grad
    out._backward = _backward

    return out

  # ------
  # re-implement all the other functions needed for the exercises below
  # your code here
  # TODO

  def __mul__(self, other):
          other = other if isinstance(other, Value) else Value(other)
          out = Value(self.data * other.data, _children=(self, other), _op=' *')
  
          def _backward():
              self.grad += other.data * out.grad
              other.grad += self.data * out.grad
  
          out._backward = _backward
          return out
  
  def __pow__(self,other):
          assert isinstance(other, (int,float)), "only supporting int/float power for now"
          out = Value(self.data**other, (self,),f'**{other}')
  
          def _backward():
              self.grad += other * (self.data ** (other-1)) * out.grad
          out._backward = _backward
          return out
  
  def __truediv__(self, other):
    other = other if isinstance(other, Value) else Value(other)
    return self * other ** -1


  def exp(self):
          out = Value(exp(self.data), (self,), 'exp')
          def _backward():
              self.grad += out.data * out.grad
          out._backward = _backward
          return out
  
  def log(self):
    out = Value(log(self.data), (self,), 'log')

    def _backward():
      self.grad += (1.0 / self.data) * out.grad
    out._backward = _backward

    return out
  
  def __neg__(self): # negative unary operator (e.g., -a)
    out = Value(-self.data, (self,), '-')

    def _backward():
      self.grad += -1.0 * out.grad
    out._backward = _backward

    return out




  # ------

  def backward(self): # exactly as in video
    topo = []
    visited = set()
    def build_topo(v):
      if v not in visited:
        visited.add(v)
        for child in v._prev:
          build_topo(child)
        topo.append(v)
    build_topo(self)

    self.grad = 1.0
    for node in reversed(topo):
      node._backward()
```

```python
# without referencing our code/video __too__ much, make this cell work
# you'll have to implement (in some cases re-implemented) a number of functions
# of the Value object, similar to what we've seen in the video.
# instead of the squared error loss this implements the negative log likelihood
# loss, which is very often used in classification.

# this is the softmax function
# https://en.wikipedia.org/wiki/Softmax_function
def softmax(logits):
  counts = [logit.exp() for logit in logits]
  denominator = sum(counts, Value(0.0))
  out = [c / denominator for c in counts]
  return out

# this is the negative log likelihood loss function, pervasive in classification
logits = [Value(0.0), Value(3.0), Value(-2.0), Value(1.0)]
probs = softmax(logits)
loss = -probs[3].log() # dim 3 acts as the label for this input example
loss.backward()
print(loss.data)

ans = [0.041772570515350445, 0.8390245074625319, 0.005653302662216329, -0.8864503806400986]
for dim in range(4):
  ok = 'OK' if abs(logits[dim].grad - ans[dim]) < 1e-5 else 'WRONG!'
  print(f"{ok} for dim {dim}: expected {ans[dim]}, yours returns {logits[dim].grad}")
```

```python
# Define logits as PyTorch tensors using the requested format
l0 = torch.tensor([0.0]).double(); l0.requires_grad = True
l1 = torch.tensor([3.0]).double(); l1.requires_grad = True
l2 = torch.tensor([-2.0]).double(); l2.requires_grad = True
l3 = torch.tensor([1.0]).double(); l3.requires_grad = True

logits_torch = [l0, l1, l2, l3]

# Softmax function implementation in PyTorch
counts = torch.stack([logit.exp() for logit in logits_torch])
denominator = counts.sum()
probs = counts / denominator

# Negative Log-Likelihood Loss for index 3
loss = -probs[3].log()

# Backpropagation
loss.backward()

# Verify gradients against expected answers
ans = [0.041772570515350445, 0.8390245074625319, 0.005653302662216329, -0.8864503806400986]

print('______________________________')
for dim in range(4):
  torch_grad = logits_torch[dim].grad.item()
  ok = 'OK' if abs(torch_grad - ans[dim]) < 1e-5 else 'WRONG!'
  print(f"{ok} for dim {dim}: expected {ans[dim]}, torch returns {torch_grad}")
```
